```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.utils import to_categorical
```

```python
fs = 1000 # frecuncia de muesteo
fc = 100 # frecuencia portada
Rs = 50 # tasa de simbolos
T = 1/Rs # tiempoo del simbolo
num_simbolos = 16 # cuantos bits se envian por se;al
t = np.arange(0, num_simbolos * T, 1/fs) # eje de tiempo
```

```python
# ASK
def generar_ask(bits, t):
    senal = np.zeros_like(t)
    largo = len(t) // len(bits)
    for i, bit in enumerate(bits):
        senal[i*largo : (i+1)*largo] = bit * np.cos(2*np.pi*fc*t[i*largo : (i+1)*largo])
    return senal

# FSK
def generar_fsk(bits, t):
    senal = np.zeros_like(t)
    largo = len(t) // len(bits)
    for i, bit in enumerate(bits):
        f_actual = fc if bit == 0 else fc + 50
        senal[i*largo : (i+1)*largo] = np.cos(2*np.pi*f_actual*t[i*largo : (i+1)*largo])
    return senal

# PSK
def generar_psk(bits, t):
    senal = np.zeros_like(t)
    largo = len(t) // len(bits)
    for i, bit in enumerate(bits):
        fase = 0 if bit == 0 else np.pi
        senal[i*largo : (i+1)*largo] = np.cos(2*np.pi*fc*t[i*largo : (i+1)*largo] + fase)
    return senal

# QAM
def generar_qam(bits, t):
    senal = np.zeros_like(t)
    num_simbolos_qam = len(bits) // 4
    largo = len(t) // num_simbolos_qam
    niveles = [-3, -1, 1, 3]

    for i in range(num_simbolos_qam):
        b = bits[i*4 : (i+1)*4]

        idx_I = b[0]*2 + b[1]
        idx_Q = b[2]*2 + b[3]

        I = niveles[idx_I]
        Q = niveles[idx_Q]

        t_local = t[i*largo : (i+1)*largo]
        senal[i*largo : (i+1)*largo] = I * np.cos(2*np.pi*fc*t_local) - Q * np.sin(2*np.pi*fc*t_local)

    return senal
```

```python
datos = []
etiquetas = []
modulaciones = ['ASK', 'FSK', 'PSK', 'QAM']
print("Genere 4000 señales limpias... (espere un momento)")

for mod in modulaciones:
    for _ in range(1000): #muestras por modulacion

        #bits al azar 0 y 1
        bits_random = np.random.randint(0, 2, num_simbolos)

        #se genera la señal
        if mod == 'ASK': onda = generar_ask(bits_random, t)
        elif mod == 'FSK': onda = generar_fsk(bits_random, t)
        elif mod == 'PSK': onda = generar_psk(bits_random, t)
        elif mod == 'QAM': onda = generar_qam(bits_random, t)

        #guardar la señal y su etiqueta
        datos.append(onda)
        etiquetas.append(mod)

#convertir a arreglos numpy
datos = np.array(datos)
etiquetas = np.array(etiquetas)
print("Dataset limpio generado, total:", len(datos), "señales.")
```

```python
#añadir Ruido AWGN
def agregar_ruido(senal, snr_db):
    potencia_senal = np.mean(senal**2)
    if potencia_senal == 0:
        potencia_senal = 1e-10 # Evita errores si es puro silencio
    potencia_ruido = potencia_senal / (10**(snr_db / 10))
    ruido = np.random.normal(0, np.sqrt(potencia_ruido), len(senal))
    return senal + ruido

# Aplicar ruido a todas las señales
datos_con_ruido = []
for i in range(len(datos)):
    snr_aleatorio = np.random.uniform(5, 20)
    onda_sucia = agregar_ruido(datos[i], snr_aleatorio)
    datos_con_ruido.append(onda_sucia)

# Se convierte a arreglo numpy
datos_con_ruido = np.array(datos_con_ruido)
print("Se añadió ruido a las 4000 señales.")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.utils import to_categorical
import numpy as np

# 1. Convertir etiquetas de texto a números
le = LabelEncoder()
etiquetas_numericas = le.fit_transform(etiquetas)

# 2. Separación 80% / 20%
X_train, X_test, y_train, y_test = train_test_split(
    datos_con_ruido, etiquetas_numericas, test_size=0.20, random_state=42
)

y_train_cat = to_categorical(y_train, num_classes=4)
y_test_cat = to_categorical(y_test, num_classes=4)

X_train = X_train[..., np.newaxis]
X_test = X_test[..., np.newaxis]

print(f"Señales para Entrenamiento: {X_train.shape}")
print(f"Etiquetas One-Hot: {y_train_cat.shape}")
```

```python
import tensorflow as tf
from tensorflow.keras import models, layers
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

print("TensorFlow versión:", tf.__version__)
```

```python
# Obtenemr la longitud de la señal
longitud_senal = X_train.shape[1]

def crear_modelo_cnn(input_shape):
    modelo = models.Sequential([
        # Capa de entrada
        layers.Input(shape=input_shape),

        # Primera capa convolucional
        layers.Conv1D(filters=32, kernel_size=3, activation='relu'),
        layers.MaxPooling1D(pool_size=2),

        # Segunda capa convolucional
        layers.Conv1D(filters=64, kernel_size=3, activation='relu'),
        layers.MaxPooling1D(pool_size=2),

        layers.Flatten(),

        # Capa densa
        layers.Dense(64, activation='relu'),
        layers.Dropout(0.3), # Ayuda a evitar el sobreajuste

        # Capa de salida
        layers.Dense(4, activation='softmax')
    ])
    return modelo

# Instanciamos el modelo
modelo = crear_modelo_cnn((longitud_senal, 1))

modelo.summary()
```

```python
modelo.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("Listo para entrenar.")
```

```python
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True,
    verbose=1
)

checkpoint = ModelCheckpoint(
    'modelo_detective_senales.keras',
    monitor='val_accuracy',
    save_best_only=True,
    verbose=1
)

# Iniciar entrenamiento
historial = modelo.fit(
    X_train, y_train_cat,
    epochs=30,               # Máximo de vueltas
    batch_size=32,           # Muestras procesadas a la vez
    validation_split=0.2,    # Usa el 20% de los datos de entrenamiento para validar
    callbacks=[early_stop, checkpoint],
    verbose=1
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))

# Gráfica de Precisión
plt.subplot(1, 2, 1)
plt.plot(historial.history['accuracy'], label='Entrenamiento')
plt.plot(historial.history['val_accuracy'], label='Validación')
plt.title('Precisión del Modelo')
plt.xlabel('Época')
plt.ylabel('Precisión')
plt.legend()
plt.grid(True)
# Grafica de Perdida
plt.subplot(1, 2, 2)
plt.plot(historial.history['loss'], label='Entrenamiento')
plt.plot(historial.history['val_loss'], label='Validación')
plt.title('Pérdida del Modelo')
plt.xlabel('Época')
plt.ylabel('Pérdida')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

print("Entrenamiento finalizado y modelo guardado")
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report
import matplotlib.pyplot as plt
import numpy as np

# Predecir con el conjunto de prueba (señales que la IA nunca vio)
y_pred = np.argmax(modelo.predict(X_test), axis=1)

# Calcular la matriz de confusión
cm = confusion_matrix(y_test, y_pred)

# Graficar la matriz
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['ASK', 'FSK', 'PSK', 'QAM'])
disp.plot(cmap='Blues', values_format='d')
plt.title('Matriz de Confusión')
plt.show()

# Imprimir reporte detallado
print(classification_report(y_test, y_pred, target_names=['ASK', 'FSK', 'PSK', 'QAM']))
```

```python
# Evaluar a la IA con un nivel específico de ruido
def evaluar_por_snr(snr_db, n_muestras=200):
    X_estres = []
    y_estres = []
    for mod_idx, mod in enumerate(['ASK', 'FSK', 'PSK', 'QAM']):
        for _ in range(n_muestras):
            bits = np.random.randint(0, 2, num_simbolos)
            if mod == 'ASK': onda = generar_ask(bits, t)
            elif mod == 'FSK': onda = generar_fsk(bits, t)
            elif mod == 'PSK': onda = generar_psk(bits, t)
            elif mod == 'QAM': onda = generar_qam(bits, t)

            # Añadir ruido con el SNR
            onda_sucia = agregar_ruido(onda, snr_db)
            X_estres.append(onda_sucia)
            y_estres.append(mod_idx)

    X_estres = np.array(X_estres)[..., np.newaxis]
    y_estres = np.array(y_estres)

    # Predecir y calcular precisión
    preds = np.argmax(modelo.predict(X_estres, verbose=0), axis=1)
    precision = np.mean(preds == y_estres)
    return precision

# Probar con distintos niveles de ruido (SNR)
niveles_snr = [-10, -5, 0, 5, 10, 15, 20]
precusiones = [evaluar_por_snr(snr) for snr in niveles_snr]
```

```python
plt.figure(figsize=(8, 5))
plt.plot(niveles_snr, precusiones, marker='o', color='red')
plt.title('Precisión del Modelo vs Nivel de Ruido (SNR)')
plt.xlabel('SNR (dB) - Nivel de Ruido')
plt.ylabel('Precisión (Accuracy)')
plt.grid(True)
plt.ylim(0, 1.05)
plt.axvline(x=0, color='gray', linestyle='--', label='Punto Crítico')
plt.legend()
plt.show()
```

```python
# Evaluar cuál modulación es la más difícil a bajo ruido
snr_malo = 0
X_bajo = []
y_bajo = []

for mod_idx, mod in enumerate(['ASK', 'FSK', 'PSK', 'QAM']):
    for _ in range(200):
        bits = np.random.randint(0, 2, num_simbolos)
        if mod == 'ASK': onda = generar_ask(bits, t)
        elif mod == 'FSK': onda = generar_fsk(bits, t)
        elif mod == 'PSK': onda = generar_psk(bits, t)
        elif mod == 'QAM': onda = generar_qam(bits, t)

        X_bajo.append(agregar_ruido(onda, snr_malo))
        y_bajo.append(mod_idx)

X_bajo = np.array(X_bajo)[..., np.newaxis]
y_bajo = np.array(y_bajo)

preds_bajo = np.argmax(modelo.predict(X_bajo, verbose=0), axis=1)

print(f"Precisión por modulación a {snr_malo} dB de ruido:")
nombres = ['ASK', 'FSK', 'PSK', 'QAM']
for i, nombre in enumerate(nombres):
    aciertos = np.mean(preds_bajo[y_bajo == i] == i)
    print(f"{nombre}: {aciertos*100:.1f}% de acierto")
```
